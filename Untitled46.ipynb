### 1 & 2: Download Correlated Data and Ensure Same Timestamp
We will download Crude Oil (`CL=F`) and 5 highly correlated assets (like XOM, CVX, SLB, COP, XLE) on a daily timeframe. Pandas automatically aligns them by timestamp.

```python
import yfinance as yf
import pandas as pd

# Crude Oil and 5 correlated assets (Energy ETF, Exxon, Chevron, Schlumberger, ConocoPhillips)
tickers = ['CL=F', 'XLE', 'XOM', 'CVX', 'SLB', 'COP']

# Download daily data (using 'Close' instead of 'Adj Close' to avoid KeyErrors in new yfinance versions)
# Fetching from 2000-01-01 to get >20 years of data
data = yf.download(tickers, start="2000-01-01", interval="1d")['Close']

# Drop NaNs to ensure the timestamp series is exactly the same across all columns
data = data.dropna()

# Save to binary Parquet format
parquet_path = 'market_data.parquet'
data.to_parquet(parquet_path)

# Load from binary Parquet format
data = pd.read_parquet(parquet_path)

display(data.head())
```

### 3: Generate Flexible Technical Indicators
Instead of hardcoding standard lengths (like RSI 14), we will generate indicators over a flexible range of lookback periods (e.g., 5 to 45).

```python
import pandas_ta as ta
import numpy as np

df = pd.DataFrame({'close': data['CL=F']})

# Create indicators with flexible ranges
lengths = [5, 10, 15, 20, 25, 30, 35, 40, 45]

for l in lengths:
    # Momentum
    df.ta.rsi(length=l, append=True)
    df.ta.cci(length=l, append=True)
    # Trend
    df.ta.sma(length=l, append=True)
    df.ta.ema(length=l, append=True)
    # Volatility
    df.ta.bbands(length=l, append=True)
    df.ta.atr(length=l, append=True)

# Target variable: 1 if next day's return is positive, else 0
df['target_return'] = df['close'].shift(-1) - df['close']
df['target_class'] = (df['target_return'] > 0).astype(int)

df = df.dropna()
print(f"Dataset shape with flexible indicators: {df.shape}")
```

### 4: Flexible Machine Learning with GPU to Determine Best Indicators
We will use XGBoost configured for GPU execution to determine which indicators carry the most predictive power.

```python
import xgboost as xgb
from sklearn.model_selection import train_test_split

# Separate features and targets
X = df.drop(columns=['close', 'target_return', 'target_class'])
y = df['target_class']

# Train/test split without shuffling to maintain time series integrity
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

# Initialize GPU-accelerated XGBoost
# NOTE: Ensure you are connected to a T4 GPU runtime in Colab (Runtime -> Change runtime type)
model = xgb.XGBClassifier(tree_method='hist', device='cuda', n_estimators=200, max_depth=5)
model.fit(X_train, y_train)

# Extract and display feature importance
importance = pd.DataFrame({
    'Indicator': X.columns,
    'Importance': model.feature_importances_
}).sort_values(by='Importance', ascending=False)

print("Top 10 Most Important Indicators:")
display(importance.head(10))
```

### 5 & 6: Backtest and Plot Delta Profit / Delta Time
We will analyze the return distributions for signals generated by the top indicators to see if they follow a normal or exponential distribution.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats

# Take the top 3 indicators for backtest analysis
top_indicators = importance['Indicator'].head(3).tolist()

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for i, ind in enumerate(top_indicators):
    # Simplified backtest delta profit calculation:
    # If indicator is above its median, assume long, else flat/short.
    # We plot the distribution of returns (delta profit / delta time) when this signal is active.
    threshold = df[ind].median()

    # Delta profit is the target return when signal condition is met
    strategy_returns = df.loc[df[ind] > threshold, 'target_return']

    sns.histplot(strategy_returns, kde=True, ax=axes[i], stat='density', bins=50)

    # Fit Normal Distribution
    mu, std = stats.norm.fit(strategy_returns)
    xmin, xmax = axes[i].get_xlim()
    x = np.linspace(xmin, xmax, 100)
    p_norm = stats.norm.pdf(x, mu, std)
    axes[i].plot(x, p_norm, 'k', linewidth=2, label=f'Normal Fit')

    # Fit Exponential (Laplace is often better for financial tails, but demonstrating fit here)
    loc, scale = stats.laplace.fit(strategy_returns)
    p_lap = stats.laplace.pdf(x, loc, scale)
    axes[i].plot(x, p_lap, 'r--', linewidth=2, label=f'Exponential/Laplace Fit')

    axes[i].set_title(f'Return Dist: {ind}')
    axes[i].legend()

plt.tight_layout()
plt.show()
```

### 8: Machine Learning Backtest and Accuracy
Let's evaluate how accurate the XGBoost model is on the unseen test data and visualize the simulated financial performance of trading based on its predictions.

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import numpy as np

# 1. Statistical Accuracy
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print(f"--- Model Performance on Test Set ---")
print(f"Accuracy: {accuracy * 100:.2f}%")
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# 2. Financial Backtest (Cumulative Returns & Risk-Adjusted Alpha)
# Get the actual price changes for the test period
test_returns = df.loc[X_test.index, 'target_return']

# Simulate a Long/Short strategy
trading_actions = np.where(y_pred == 1, 1, -1)

# Reverse strategy for 2020 and above
is_2020_or_later = X_test.index.year >= 2020
trading_actions = np.where(is_2020_or_later, -trading_actions, trading_actions)

strategy_returns = trading_actions * test_returns

# Calculate Sharpe Ratio (Alpha / Risk-Adjusted Return)
daily_mean = strategy_returns.mean()
daily_std = strategy_returns.std()
sharpe_ratio = np.sqrt(252) * (daily_mean / daily_std) if daily_std > 0 else 0
print(f"Strategy Sharpe Ratio (Test Set, Reversed >= 2020): {sharpe_ratio:.2f}")

# Calculate cumulative sum of returns (points gained/lost)
cumulative_market = test_returns.cumsum()
cumulative_strategy = strategy_returns.cumsum()

# Plot the backtest
plt.figure(figsize=(14, 6))
plt.plot(cumulative_market.index, cumulative_market, label='Buy & Hold (Market)', color='gray', alpha=0.7)
plt.plot(cumulative_strategy.index, cumulative_strategy, label=f'ML Strategy (Sharpe: {sharpe_ratio:.2f})', color='blue', linewidth=2)
plt.title('Out-of-Sample Backtest: ML Strategy vs Buy & Hold (Reversed >= 2020)')
plt.xlabel('Date')
plt.ylabel('Cumulative Profit (Delta Points)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

### 10: Historical Options Data for ML Pipeline
Since we lack historical options data from free APIs, we will simulate a "Historical Max Pain" proxy by rounding a 20-day Simple Moving Average to the nearest integer. This mimics the "magnet" effect of strike prices. We then feed the distance between the close price and this  Max Pain into our XGBoost model.

```python
import xgboost as xgb
from sklearn.model_selection import train_test_split
import pandas as pd

# 1. Create a MOCK Historical Max Pain feature (For demonstration purposes only)
# We simulate strike prices by rounding a moving average to the nearest whole number
df['Mock_Max_Pain'] = df['close'].rolling(window=20).mean().round()

# Feature: How far is the current price from the "Max Pain" strike?
df['Price_to_Max_Pain_Dist'] = df['close'] - df['Mock_Max_Pain']

# Drop NaNs generated by the new rolling window
df_ml = df.dropna()

# 2. Setup Machine Learning with the new feature
X_opt = df_ml.drop(columns=['close', 'target_return', 'target_class', 'Mock_Max_Pain'])
y_opt = df_ml['target_class']

X_train_o, X_test_o, y_train_o, y_test_o = train_test_split(X_opt, y_opt, test_size=0.2, shuffle=False)

print("Training XGBoost with simulated Options Data...")
model_opt = xgb.XGBClassifier(tree_method='hist', device='cuda', n_estimators=200, max_depth=5)
model_opt.fit(X_train_o, y_train_o)

# 3. Check if the model finds the "Max Pain Distance" useful
importance_opt = pd.DataFrame({
    'Indicator': X_opt.columns,
    'Importance': model_opt.feature_importances_
}).sort_values(by='Importance', ascending=False)

print("\nTop 10 Features (Including Mock Max Pain):")
display(importance_opt.head(10))
```

### 11: Calculate Real Current Max Pain
Here is the actual mathematical calculation for **Max Pain**. We apply this to the live options chain we downloaded earlier for `XLE`. If you save this output daily to a database, you can build your own historical options dataset.

```python
def calculate_max_pain(calls_df, puts_df):
    # Get all unique strike prices available in the chain
    strikes = sorted(set(calls_df['strike']).union(set(puts_df['strike'])))

    max_pain_strike = 0
    min_pain_value = float('inf')

    # Test every strike price to see which one causes the least total payout (pain to option buyers)
    for test_strike in strikes:
        total_pain = 0

        # Call buyers make money if Expiry Price > Call Strike
        for _, call in calls_df.iterrows():
            if test_strike > call['strike']:
                total_pain += (test_strike - call['strike']) * call['openInterest'] * 100 # Multiplier

        # Put buyers make money if Expiry Price < Put Strike
        for _, put in puts_df.iterrows():
            if test_strike < put['strike']:
                total_pain += (put['strike'] - test_strike) * put['openInterest'] * 100

        # Find the strike with the lowest total payout
        if total_pain < min_pain_value:
            min_pain_value = total_pain
            max_pain_strike = test_strike

    return max_pain_strike, min_pain_value

if 'calls' in locals() and 'puts' in locals():
    current_max_pain, pain_dollar_value = calculate_max_pain(calls, puts)

    print(f"--- REAL LIVE OPTIONS DATA ---")
    print(f"Ticker: {ticker_symbol}")
    print(f"Nearest Expiry: {nearest_expiry}")
    print(f"\n🎯 CURRENT MAX PAIN STRIKE: ${current_max_pain}")
    print(f"Minimum Option Value Payout: ${pain_dollar_value:,.2f}")
else:
    print("Options chain not found. Please re-run the previous cell fetching options data.")
```

### 12: Integrating Real Historical Options-Derived Volatility Data
Instead of a simulated Max Pain proxy, we will integrate `^OVX` (Crude Oil Volatility Index) and `^VIX` (S&P 500 Volatility Index). These are real historical metrics calculated directly from options premiums, representing the market's expected volatility and fear/hedging demand.

```python
import yfinance as yf
import xgboost as xgb
from sklearn.model_selection import train_test_split
import pandas as pd

print("Downloading Real Options-Derived Volatility Data (^OVX and ^VIX)...")
# Download VIX and OVX (Crude Oil Volatility Index)
opt_vix_data = yf.download(['^OVX', '^VIX'], start="2007-05-10", interval="1d")['Close']
opt_vix_data = opt_vix_data.dropna()

# Ensure we are using the main dataframe without the previous mock data
df_real_opt = df.copy()
if 'Mock_Max_Pain' in df_real_opt.columns:
    df_real_opt = df_real_opt.drop(columns=['Mock_Max_Pain', 'Price_to_Max_Pain_Dist'])

# Inner join to align timestamps (OVX data starts around 2007, so we will truncate older data)
df_merged = df_real_opt.join(opt_vix_data, how='inner')

# Add momentum features for the options data (how fast is fear rising/falling?)
df_merged['OVX_Change'] = df_merged['^OVX'].pct_change()
df_merged['VIX_Change'] = df_merged['^VIX'].pct_change()

df_merged = df_merged.dropna()
print(f"Merged Dataset shape with Real Options Volatility: {df_merged.shape}")

# Setup Machine Learning Features
X_real = df_merged.drop(columns=['close', 'target_return', 'target_class'])
y_real = df_merged['target_class']

# Train/Test Split (Time Series)
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(X_real, y_real, test_size=0.2, shuffle=False)

print("\nTraining XGBoost with Real Options-Implied Volatility Data...")
model_real = xgb.XGBClassifier(tree_method='hist', device='cuda', n_estimators=200, max_depth=5)
model_real.fit(X_train_r, y_train_r)

# Check Feature Importance
importance_real = pd.DataFrame({
    'Indicator': X_real.columns,
    'Importance': model_real.feature_importances_
}).sort_values(by='Importance', ascending=False)

print("\nTop 15 Features (Including Real Options Volatility):")
display(importance_real.head(15))
```

### 13: Evaluate Model Performance with Real Options Data
Let's run the backtest for the model trained with real options implied volatility data to see its Out-Of-Sample accuracy and risk-adjusted returns (Sharpe Ratio).

```python
from sklearn.metrics import accuracy_score, classification_report
import matplotlib.pyplot as plt
import numpy as np

# 1. Statistical Accuracy
y_pred_r = model_real.predict(X_test_r)
accuracy_r = accuracy_score(y_test_r, y_pred_r)

print(f"--- Model Performance with Real Options Data (Test Set) ---")
print(f"Accuracy: {accuracy_r * 100:.2f}%")
print("\nClassification Report:")
print(classification_report(y_test_r, y_pred_r))

# 2. Financial Backtest (Cumulative Returns & Risk-Adjusted Alpha)
# Get the actual price changes for the test period
test_returns_r = df_merged.loc[X_test_r.index, 'target_return']

# Simulate a Long/Short strategy
trading_actions_r = np.where(y_pred_r == 1, 1, -1)

# Reverse strategy for 2020 and above (to match our previous logic)
is_2020_or_later_r = X_test_r.index.year >= 2020
trading_actions_r = np.where(is_2020_or_later_r, -trading_actions_r, trading_actions_r)

strategy_returns_r = trading_actions_r * test_returns_r

# Calculate Sharpe Ratio (Alpha / Risk-Adjusted Return)
daily_mean_r = strategy_returns_r.mean()
daily_std_r = strategy_returns_r.std()
sharpe_ratio_r = np.sqrt(252) * (daily_mean_r / daily_std_r) if daily_std_r > 0 else 0
print(f"Strategy Sharpe Ratio (Real Options, Reversed >= 2020): {sharpe_ratio_r:.2f}")

# Calculate cumulative sum of returns (points gained/lost)
cumulative_market_r = test_returns_r.cumsum()
cumulative_strategy_r = strategy_returns_r.cumsum()

# Plot the backtest
plt.figure(figsize=(14, 6))
plt.plot(cumulative_market_r.index, cumulative_market_r, label='Buy & Hold (Market)', color='gray', alpha=0.7)
plt.plot(cumulative_strategy_r.index, cumulative_strategy_r, label=f'Real Options Strategy (Sharpe: {sharpe_ratio_r:.2f})', color='purple', linewidth=2)
plt.title('Out-of-Sample Backtest: ML Strategy + Real Options Volatility Data')
plt.xlabel('Date')
plt.ylabel('Cumulative Profit (Delta Points)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

### 15: Live Prediction for Today (Without Dropping NaN Targets)
To predict using today's live data, we need to rebuild the features for the last few days and extract the final row *before* any targets are calculated. This ensures today's data isn't dropped.

```python
import yfinance as yf
import pandas as pd
import pandas_ta as ta

print("Fetching live data to include today...")
# 1. Fetch recent data (3 months to ensure 45-day indicators compute properly)
recent_cl = yf.download('CL=F', period="3mo", interval="1d")['Close']
if isinstance(recent_cl, pd.DataFrame):
    recent_cl = recent_cl.iloc[:, 0]

df_today = pd.DataFrame({'close': recent_cl})

# 2. Compute technical indicators
lengths = [5, 10, 15, 20, 25, 30, 35, 40, 45]
for l in lengths:
    df_today.ta.rsi(length=l, append=True)
    df_today.ta.cci(length=l, append=True)
    df_today.ta.sma(length=l, append=True)
    df_today.ta.ema(length=l, append=True)
    df_today.ta.bbands(length=l, append=True)
    df_today.ta.atr(length=l, append=True)

# 3. Fetch recent Options Volatility data
recent_vix_ovx = yf.download(['^OVX', '^VIX'], period="3mo", interval="1d")['Close']
df_today = df_today.join(recent_vix_ovx, how='inner')

# 4. Compute Volatility Momentum
df_today['OVX_Change'] = df_today['^OVX'].pct_change()
df_today['VIX_Change'] = df_today['^VIX'].pct_change()

# 5. Extract ONLY the features expected by the model for the absolute latest day
# We fill any slight forward NAs in case VIX/OVX hasn't updated yet today
df_today = df_today.ffill()
today_features = df_today[X_real.columns].iloc[[-1]]

# 6. Predict using the real-options trained model
pred_today = model_real.predict(today_features)
pred_proba_today = model_real.predict_proba(today_features)

direction = "UP (Positive Return)" if pred_today[0] == 1 else "DOWN (Negative or Zero Return)"

print(f"\n--- LATEST LIVE PREDICTION ---")
print(f"Data Date: {today_features.index[0].date()}")
print(f"Predicted Next Price Movement (CL=F): {direction}")
print(f"Probability of UP: {pred_proba_today[0][1]:.4f}")
print(f"Probability of DOWN: {pred_proba_today[0][0]:.4f}")
```
